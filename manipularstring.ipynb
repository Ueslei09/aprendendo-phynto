**Operação Matematica**

```python
a = 5
b = 3
print(f'soma: {a+b}')
print(f'subtração: {a-b}')
print(f'multiplicação: {a*b}')
print(f'divisão: {round(a/b,2)}')
print(f'resto da divisão: {a%b}')
print(f'potencia: {a**b}')
print(f'raiz quadrada: {round(a**(1/2),2)}')
```
